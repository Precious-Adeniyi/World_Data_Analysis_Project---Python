# World Data Analysis Using Python

## Project Overview

### This project uses the REST Countries API to retrieve and analyze country level data. The analysis includes population density calculations, summary statistics, and country-level insights.

```python
# Import required libraries
import requests
import statistics
```

## Data Collection

### Data is retrieved from the REST Countries API using Python and the requests library.

```python
# Retrieve data from the REST Countries API
api_url = "https://restcountries.com/v3.1/all?fields=name,population,area"
api_response = requests.get(api_url)
data = api_response.json()
```

## Population Density Analysis

### Population density is calculated as population divided by land area for each country.

```python
# Calculate population density for each country

pop_densities = []

for country_data in data:

    country_name = country_data['name']['common']
    pop = country_data['population']
    area = country_data['area'] if 'area' in country_data else None

    if area is not None and area > 0:
        pop_density = pop / area
        pop_densities.append(pop_density)
    else:
        pop_density = None

print(f"{country_name}: {pop_density}")
```

## Statistical Analysis

### The mean, median, and standard deviation of population density are calculated to understand the distribution of population across countries.

```python
# Calculate the mean, median and standard deviation of the population density.

mean_density = round(statistics.mean(pop_densities), 2)
median_density = round(statistics.median(pop_densities),2)
std_dev_density = round(statistics.stdev(pop_densities),2)


print(f"The Mean Population Density: {mean_density}")
print(f"The Median Population Density: {median_density}")
print(f"The Standard Deviation of Population Density: {std_dev_density}")
```

```python
# Calcuate the number of countries who are UN members, and the number of countries who use the Euro as a currency.

No_of_UN_members = sum(1 for country in data if country.get('unMember'))
No_of_euro_users = sum(1 for country in data if 'eur' in country.get('currencies', {}))

print(f"The Number of UN Member Countries: {No_of_UN_members}")
print(f"The Number of Countries Using Euro: {No_of_euro_users}")
```

```python
# This code reads numbered word pairs, builds a dictionary, then selects specific numbers based on a growing pyramid pattern, and uses those to reveal a hidden sentence 
def decode(message_file):
    with open(message_file, 'r', encoding='utf-8') as file:
        lines = file.readlines()

    # Extract numbers and corresponding words
    number_word_pairs = [line.strip().split() for line in lines]
    number_word_dict = {int(pair[0]): pair[1] for pair in number_word_pairs}

    # Identify the numbers at the end of each pyramid line
    pyramid_end_numbers = []
    step, count = 1, 0
    for number in sorted(number_word_dict.keys()):
        count += 1
        if count == step:
            pyramid_end_numbers.append(number)
            step += 1

    # Construct the decoded message
    decoded_message = ' '.join([number_word_dict[number] for number in pyramid_end_numbers])

    return decoded_message

# Example usage
message = decode(r'C:\Users\preci\Downloads\Jupyter\file_path.txt')
print(message)
```
